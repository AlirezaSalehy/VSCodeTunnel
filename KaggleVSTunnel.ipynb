## Tunnel
- Sets up a tunnel to connet kaggle run-time to your local VS Code
- It first authenticates you by logging into your Microsoft or Github account
- Then in local VS Code, select `Connect to Tunnel`, after authentication select the `kaggle`
- Note that the new arguements (`cli-data-dir`, `server-data-dir`, `extensions-dir`) are optional but very convenient in order not to reinstall extensions or not to reconfigure workspace settings on every runtime 

```python
# Creating directories if not exist for cli-data-dir, server-data-dir, extensions-dir
BASE="/kaggle/working/.vscode_data"
!mkdir -p $BASE/cli $BASE/server $BASE/ext
```

```python
!./vscode-cli/code --version
!./vscode-cli/code tunnel user login --provider github
!./vscode-cli/code tunnel --accept-server-license-terms --name kaggle \
                          --server-data-dir $BASE/server \
                          --extensions-dir $BASE/ext 
```
